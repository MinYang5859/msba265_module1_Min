```python
print("====================================================")
print("          UNIVARIATE SKEWNESS DIAGNOSTICS           ")
print("====================================================")

numeric_df = df.select_dtypes(include=[np.number])
skewness_series = numeric_df.skew().sort_values(ascending=False)

skew_summary = pd.DataFrame({
    "Feature": skewness_series.index,
    "Skewness Coefficient": skewness_series.values,
    "Diagnosis": [
        "Extremely Right-Skewed"
        if s > 3.0
        else "Moderately Skewed"
        if abs(s) > 1.0
        else "Unskewed / Symmetric"
        for s in skewness_series.values
    ],
    "Action Required": [
        "Log / Box-Cox Transform"
        if abs(s) > 1.0
        else "Standard Scaler Sufficient"
        for s in skewness_series.values
    ]
})

display(skew_summary)
```

##  Skewness Diagnosis and Domain Override

`ClaimNb` and `Density` show the strongest positive skewness. However, the same transformation rule should not be applied automatically to both variables.

`Density` is a non-negative continuous feature. A `log1p(Density)` transformation may reduce the influence of the long right tail while retaining valid urban policyholders.

In contrast, `ClaimNb` is a discrete count variable with many zero values. Its high skewness reflects the natural structure of insurance claim frequency rather than a simple data quality problem. Therefore, the automatic Log/Box-Cox recommendation is overridden. `ClaimNb` should be modeled as a count or rate target, such as with a Poisson or Negative Binomial model, while accounting for `Exposure`.

`IDpol` is excluded from transformation decisions because it is an identifier rather than a meaningful continuous predictor.

```python
print("====================================================")
print("          PEARSON MULTICOLLINEARITY AUDIT           ")
print("====================================================")

corr_matrix = numeric_df.corr().abs()

upper_tri = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

high_corr_pairs = []

for col in upper_tri.columns:
    high_corr_features = upper_tri.index[
        upper_tri[col] >= 0.85
    ].tolist()

    for feat in high_corr_features:
        val = upper_tri.loc[feat, col]
        high_corr_pairs.append({
            "Feature 1": feat,
            "Feature 2": col,
            "Pearson Correlation (|r|)": round(val, 4),
            "Status": "ACTION REQUIRED: Severe Multicollinearity"
        })

if high_corr_pairs:
    display(pd.DataFrame(high_corr_pairs))
else:
    print(
        "[+] No severe collinear feature pairs (|r| >= 0.85) "
        "detected across numerical metrics."
    )
```

##  Collinearity Audit

No numerical feature pair reached the severe Pearson correlation threshold of `|r| ≥ 0.85`. Therefore, the audit did not identify an immediate need to remove a numerical predictor because of severe pairwise linear correlation.

This result only evaluates pairwise Pearson correlation. It does not prove that all Variance Inflation Factors are low, and it does not capture nonlinear relationships.

In a formal predictor audit, `IDpol` should be excluded because it is an identifier, and `ClaimNb` should be treated as the target rather than an input predictor.

##  Visual Distribution Audit

### Population Density

`Density` is strongly right-skewed. Its median is 393 people per square kilometer, while its mean is approximately 1,792.42. The maximum reaches 27,000, showing that a relatively small number of high-density municipalities extend the right tail.

These values may represent valid urban policyholders rather than data errors. They should not be removed automatically. A `log1p(Density)` transformation may be considered for models that are sensitive to skewness.

### Driver Age

`DrivAge` is approximately balanced with mild positive skewness. Its first quartile is 34, median is 44, third quartile is 55, and observed range is 18 to 100 years.

The distribution does not require a logarithmic transformation or automatic outlier removal. Standard scaling may be considered for scale-sensitive models.

### BonusMalus

`BonusMalus` is concentrated near its lower boundary. Its first quartile and median are both 50, while the third quartile is 64.

Values above the Tukey upper fence are statistical outliers, but they may represent legitimate higher-risk or surcharge policyholders. These records should not be deleted automatically. A binary indicator such as `BonusMalus > 100` may provide an interpretable modeling feature.

### Vehicle Age

`VehAge` is positively skewed, with most vehicles concentrated at lower ages and a sparse right tail extending to 100 years. Its first quartile is 2, median is 6, and third quartile is 11.

Values above approximately 24.5 years satisfy the Tukey definition of statistical outliers. Extremely old vehicle ages require additional domain validation before deciding whether to retain, cap, or remove them.

```python
target_feature = "Density"

Q1 = df[target_feature].quantile(0.25)
Q3 = df[target_feature].quantile(0.75)
IQR = Q3 - Q1

iqr_lower = Q1 - 1.5 * IQR
iqr_upper = Q3 + 1.5 * IQR

df_iqr_cleaned = df[
    (df[target_feature] >= iqr_lower)
    & (df[target_feature] <= iqr_upper)
]

z_scores = (
    df[target_feature] - df[target_feature].mean()
) / df[target_feature].std()

df_z_cleaned = df[abs(z_scores) <= 3.0]

print(f"=== OUTLIER AUDIT REPORT FOR {target_feature} ===")
print(f"Original Record Count: {len(df):,} records")
print(
    f"Tukey IQR Range [{iqr_lower:.2f}, {iqr_upper:.2f}]: "
    f"Excludes {len(df) - len(df_iqr_cleaned):,} records"
)
print(
    f"Z-Score Threshold (|Z| <= 3.0): "
    f"Excludes {len(df) - len(df_z_cleaned):,} records"
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(
    x=df[target_feature],
    ax=axes[0],
    color="salmon"
)
axes[0].set_title(
    f"BEFORE Filtering: Raw {target_feature}",
    fontsize=12,
    fontweight="bold"
)
axes[0].set_xlabel("Population Density (People / km2)")

sns.boxplot(
    x=df_iqr_cleaned[target_feature],
    ax=axes[1],
    color="lightgreen"
)
axes[1].set_title(
    f"AFTER Filtering: IQR Tukey Filtered {target_feature}",
    fontsize=12,
    fontweight="bold"
)
axes[1].set_xlabel("Population Density (People / km2)")

plt.tight_layout()

fig_filter_path = os.path.join(
    "..",
    "reports",
    "figures",
    "outlier_filtering_comparison.png"
)

plt.savefig(fig_filter_path, dpi=300)
plt.show()

print(
    f"[+] Saved visual outlier comparison artifact to: "
    f"{fig_filter_path}"
)
```

##  Tukey IQR and Z-Score Outlier Comparison

### Baseline Outlier Audit

The outlier audit evaluated `Density` across 678,013 policy records. The Tukey IQR method established a valid range from -2,257 to 4,007 people per square kilometer and flagged 77,566 records, or approximately 11.44%, as statistical outliers.

The Z-score method flagged 15,209 records, or approximately 2.24%. Therefore, the Tukey method identified 62,357 more observations than the Z-score method.

### Methodological Interpretation

The two methods produce different results because `Density` is strongly right-skewed. Extreme metropolitan values increase both the mean and standard deviation, causing the parametric Z-score cutoff to move farther into the right tail.

The Tukey IQR method uses quartiles and is less affected by extreme values. It therefore identifies more high-density observations as statistical outliers.

### Business Decision

Statistical outliers are not automatically data errors. Observations above 4,007 people per square kilometer may represent valid urban policyholders.

If the business objective is statewide or national insurance pricing, automatically removing 11.44% of the records could introduce geographic selection bias. In this scenario, the records should be retained and `log1p(Density)` should be evaluated.

If the downstream model is highly sensitive to leverage points and the analysis is intentionally limited to a lower-density population, Tukey filtering may be considered only with a documented business justification.

For tree-based models, mathematical outlier removal may be unnecessary because these models split features using decision thresholds and are generally less sensitive to monotonic transformations.
